```python
origtex = list(text)
#origtex.count('i')
origtex
```

```python
token_pairs = dict()

for i in range(len(origtex)-1):
    pair = origtex[i], origtex[i+1]
    if pair in token_pairs:
        token_pairs[pair] += 1
    else:
        token_pairs[pair] = 1
token_pairs        
```

```python
mostFrequentIdx = np.argmax(list(token_pairs.values()))
mostFrequentPair = list(token_pairs.keys())[mostFrequentIdx]

print(f"The most frequent pair is {mostFrequentPair} with {token_pairs[mostFrequentPair]} occurrences.")
```

```python
vocab[mostFrequentPair] = max(vocab.values()) + 1
vocab 
```

```python
new_text = []
i = 0
while i < len(origtex):
    if i < len(origtex)-1 and (origtex[i], origtex[i+1]) == mostFrequentPair:
        new_text.append(mostFrequentPair[0] + mostFrequentPair[1])   # ' h'   # now a string like ' h'
        i += 2
    else:
        new_text.append(origtex[i])
        i += 1
print(new_text)
```

# Writing it as a function

```python
# > ------------------------------------
def get_pair_stats(text2pair):
  token_pairs = dict()

  # loop over tokens
  for i in range(len(text2pair)-1):

    # create a pair
    pair = text2pair[i] + text2pair[i+1]

    # increase pair frequencies
    if pair in token_pairs:
      token_pairs[pair] += 1
    else:
      token_pairs[pair] = 1

  return token_pairs
```

```python
# > ----------------------------------
def update_vocab(token_pairs,vocab):

  # find the most frequent pair
  idx = np.argmax(list(token_pairs.values()))
  newtok = list(token_pairs.keys())[idx]

  # update the vocab
  vocab[newtok] = max(vocab.values())+1
  return vocab,newtok
# ---------------------------------- <
```

```python
# ----------------------------------
def generate_new_token_seq(prevtext,newtoken):

  # initialize a new text list
  newtext = []

  # loop through the list
  i = 0
  while i<(len(prevtext)-1):

    # test whether the pair of this and the following elements match the newly-crea
    if (prevtext[i]+prevtext[i+1]) == newtoken:
      newtext.append(newtoken)
      i += 2 # skip the next character

    # not a pair
    else:
      newtext.append(prevtext[i])
      i += 1 # move to the next character
      
  return newtext
```
