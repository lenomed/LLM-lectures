# create a rapper name with you first-name last name and favorite colour

```python
my_string = "Sunday Jethro Black"

tokens = tokenizer.encode(my_string)
sorted_tokens = sorted(tokens)

print(f'Original tokens: {tokens}')
print(f'Sorted tokens:   {sorted_tokens}')

# see what each sorted token represents
for t in sorted_tokens:
  print(f'{t:>6} = "{tokenizer.decode([t])}"')
```

```python
import numpy as np

first_name = 'Sunday'
last_name  = 'Jethro'
color      = 'Black'
my_string  = f'{first_name} {last_name} {color}'

# tokenize and sort ascending (previous challenge)
sorted_tokens = sorted(tokenizer.encode(my_string))

# 1. token ID of first letter of first name, minus 1
part1 = tokenizer.encode(first_name[0])[0] - 1

# 2. token ID of last letter of last name, plus 1
part2 = tokenizer.encode(last_name[-1])[0] + 1

# 3. first and last sorted tokens, plus the average of the middle ones
part3_first  = sorted_tokens[0]
part3_last   = sorted_tokens[-1]
part3_middle = int(round(np.mean(sorted_tokens[1:-1])))

rapper_ids = [part1, part2, part3_first, part3_middle, part3_last]

for t in rapper_ids:
  print(f'{t:>6} = "{tokenizer.decode([t])}"')

rapper_name = ''.join(tokenizer.decode([t]) for t in rapper_ids).replace(' ', '')
print(f'\nYour rapper name is: {rapper_name}')
```

```python
first_name = 'Sunday'
second_name = 'Jethro'
favorite_color = 'Black'

text = " ".join([first_name, second_name, favorite_color])
text
```

```python
idx = tokenizer.encode(text)
idx.sort()
print(f'idx = {idx}, \n decoded "{tokenizer.decode(idx)}"')
```

```python
def not_normal(token_str):
  pattern = re.compile(r'^[A-Za-z0-9\s]+$')
  return not bool(pattern.fullmatch(token_str))
```

```python
# part 1: first letter of first name -1
part1 = tokenizer.encode(first_name[0])[0]-1
while not_normal(tokenizer.decode([part1])):
  part1 += 1
```

```python
# part 2: last letter of the last name +1
part2 = tokenizer.encode(last_name[-1])[0]+1
while not_normal(tokenizer.decode([part2])):
  part2 += 1
```

```python
# part 3: last/first + average of middle
part3 = int(idx[-1]/idx[0] + np.mean(idx[1:-1]))
while not_normal(tokenizer.decode([part3])):
  part3 += 1
```

```python
rappername = [part1,part2,part3]
print(f'My rapper name is "{tokenizer.decode(rappername)}"')
```
