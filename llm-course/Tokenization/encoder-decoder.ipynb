```python
import re
import numpy as np
```

```python
def encoder(text):
    tokens = re.split(r'\s+', text)
    return [(i, word) for i, word in enumerate(tokens)]
```

```python
encoder('hello world, how are you?')
```

```python
def decoder(idx):
    return [word for i, word in idx]
```

```python
test = ['we are the result of what we have learned', 'either way we need to be very sober']
allwords = re.split(r'\s+', ' '.join(test))
vocab = sorted(list(set(allwords)))
print(vocab)
enc = encoder(test[0])
print(enc)
print(' '.join(decoder(enc)))
```

```python
import matplotlib
import matplotlib.pyplot as plt
```

```python
allwords = ' '.join(test)
print(allwords)
```

```python
tokens = encoder(allwords)
```

```python
print(type(tokens), type(tokens[0]), tokens[:3])
```

```python
ids = [vocab.index(word) for i, word in tokens]
_,ax = plt.subplots(figsize=(12, 5))
ax.plot(ids,'ks', markersize=10, markerfacecolor=(0.7, 0.7, 0.9))
ax.set_xlabel('word index'), ax.set_yticks(range(len(vocab)))
ax.grid(linestyle='--', axis='y')

ax2 = ax.twinx()
ax2.plot(ids, alpha =0)
ax2.set_yticks(range(len(vocab)), labels=vocab)
plt.show()
```
