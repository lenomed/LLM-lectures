```python
from transformers import GPT2TokenizerFast

claudeTokenizer = GPT2TokenizerFast.from_pretrained('Xenova/claude-tokenizer')
```

```python
claudeTokenizer.vocab_size
```

```python
claudeTokenizer.encode('this is some text')
```

```python
claudeTokenizer('this is some text')
```

```python
claudeTokenizer('this is some text', return_tensors='pt')
```

```python
words = [ 'hypothetical',
          ' hypothetical',
          'hypothetical ',
          ' hypothteical'  ]

for word in words:
  toks = claudeTokenizer.encode(word)
  print(f'"{word}" has {len(toks)} tokens ({toks}):\n   {[claudeTokenizer.decode(t) for t in toks]}')
```
