```python
!pip install tiktoken

import tiktoken

gpt4tokenizer = tiktoken.get_encoding("cl100k_base")

from transformers import BertTokenizer

bertTokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
```

```python
def bertToGPT4(bertTok):
    decoded_bert = bertTokenizer.decode(bertTok)
    return gpt4tokenizer.encode(decoded_bert)
```

```python
def gpt4ToBert(gpt4Tok):
    decoded_gpt4 = gpt4tokenizer.decode(gpt4Tok)
    return bertTokenizer.encode(decoded_gpt4)
```

```python
text = "Hello, how are you?"

bert_tokens = bertTokenizer.encode(text)
gpt4_tokens = gpt4tokenizer.encode(text)

print(f' {bertToGPT4(bert_tokens)}')
print(f' {gpt4tokenizer.decode(bertToGPT4(bert_tokens))}')

print(f'{gpt4ToBert(gpt4_tokens)}')

print(f' {bertTokenizer.decode(gpt4ToBert(gpt4_tokens))}')
```

```python
txt = "i wanted to paste in a thought-provoking quote here, but i didn't."

b_encoder = bertTokenizer.encode(txt)[1:-1]
print(b_encoder)
g_encoder = gpt4tokenizer.encode(txt)
print(f'\n this is bert to gpt4 txt: {bertToGPT4(b_encoder)}')
print(f'\n this are the decoded tokens from bertToGPT4: {gpt4tokenizer.decode(bertToGPT4(b_encoder))}')
print(f'\n this is gpt4 to bert txt: {gpt4ToBert(g_encoder)}')
print(f'\n conver the gpt4 back to bert {bertTokenizer.decode(gpt4ToBert(g_encoder))}')
```
