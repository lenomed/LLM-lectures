```python
import numpy as np
# some text with lots of repetitions
text = 'walk walks walked walking talk talks talked talking the then there them they walker talker thinker thinking think thinks thought a walking talker talks while a talking walker walks then the thinker thinks they walk then they talk then they think the walker walked there and the talker talked there the thinker thinking there thought them walking and talking and thinking walks talks thinks walked talked thought the walking talker and the talking walker then walk there and think about them while they talk the walker thinks the talker walks the thinker talks then they walk and talk and think there while the walking talking thinking walkers talkers thinkers walked talked thought again and again and again then the walker talker thinker thinks talks walks them there'
chars = list(set(text))
chars.sort() # initial vocab is sorted

for l in chars:
  print(f'"{l}" appears {text.count(l)} times.')
```

```python
# make a vocabulary
vocab = { word:i for i,word in enumerate(chars) }
vocab
```

```python
# the text needs to be a list, not a string
# each element in the list is a token
origtext = list(text)
print(text)
print(origtext)
```

```python
# > ------------------------------------
def get_pair_stats(text2pair):
  token_pairs = dict()

  # loop over tokens
  for i in range(len(text2pair)-1):

    # create a pair
    pair = text2pair[i] + text2pair[i+1]

    # increase pair frequencies
    if pair in token_pairs:
      token_pairs[pair] += 1
    else:
      token_pairs[pair] = 1

  return token_pairs
```

```python
# > ----------------------------------
def update_vocab(token_pairs,vocab):

  # find the most frequent pair
  idx = np.argmax(list(token_pairs.values()))
  newtok = list(token_pairs.keys())[idx]

  # update the vocab
  vocab[newtok] = max(vocab.values())+1
  return vocab,newtok
# ---------------------------------- <
```

```python
# ----------------------------------
def generate_new_token_seq(prevtext,newtoken):

  # initialize a new text list
  newtext = []

  # loop through the list
  i = 0
  while i<(len(prevtext)-1):

    # test whether the pair of this and the following elements match the newly-crea
    if (prevtext[i]+prevtext[i+1]) == newtoken:
      newtext.append(newtoken)
      i += 2 # skip the next character

    # not a pair
    else:
      newtext.append(prevtext[i])
      i += 1 # move to the next character
      
  return newtext
```

```python
# re-initialize the vocab (again...)
vocab = { word:i for i,word in enumerate(chars) }
```

```python
# how many tokens in the vocab?
vocab_size = 50

# make a copy of the original text
updated_text = origtext.copy()

while len(vocab)<vocab_size:

  # get pair statistics
  pairs = get_pair_stats(updated_text)

  # update the dictionary
  vocab,newtoken = update_vocab(pairs,vocab)

  # get a new list of tokens
  updated_text = generate_new_token_seq(updated_text,newtoken)
  print(f'Vocab has {len(vocab)} tokens after adding "{newtoken}"')
```
